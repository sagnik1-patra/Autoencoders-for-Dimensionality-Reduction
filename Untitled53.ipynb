```python
# =====================================================================
# AUTOENCODERS FOR DIMENSIONALITY REDUCTION
# MNIST IMAGE COMPRESSION + DENOISING
# FULL CLEAN FIXED VERSION
# =====================================================================

import os
import json
import time
import pickle
import shutil
import ssl
import urllib.request
import warnings

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from skimage.metrics import peak_signal_noise_ratio
from skimage.metrics import structural_similarity

try:
    import yaml
except ImportError:
    raise ImportError("Install PyYAML first: pip install pyyaml")


# =====================================================================
# 1. CONFIGURATION
# =====================================================================

PROJECT_DIR = r"C:\Users\sagni\Downloads\Autoencoders for Dimensionality Reduction"

os.makedirs(PROJECT_DIR, exist_ok=True)

SEED = 42
MAX_EPOCHS = 50
BATCH_SIZE = 256
LEARNING_RATE = 0.001
VALIDATION_SPLIT = 0.10

VANILLA_LATENT_DIM = 32
CONV_LATENT_DIM = 16
VAE_LATENT_DIM = 20
VAE_BETA = 1.0

LATENT_VIS_SAMPLES = 3000

np.random.seed(SEED)
tf.random.set_seed(SEED)


def output_path(filename):
    return os.path.join(PROJECT_DIR, filename)


print("=" * 70)
print("AUTOENCODERS FOR DIMENSIONALITY REDUCTION")
print("=" * 70)
print("Project directory:")
print(PROJECT_DIR)


# =====================================================================
# 2. SAVE CONFIGURATION
# =====================================================================

config = {
    "project": "Autoencoders for Dimensionality Reduction",
    "dataset": "MNIST",
    "seed": SEED,
    "epochs": MAX_EPOCHS,
    "batch_size": BATCH_SIZE,
    "learning_rate": LEARNING_RATE,
    "validation_split": VALIDATION_SPLIT,
    "vanilla_latent_dimension": VANILLA_LATENT_DIM,
    "convolutional_latent_dimension": CONV_LATENT_DIM,
    "vae_latent_dimension": VAE_LATENT_DIM,
    "gaussian_noise_factor": 0.5,
    "vae_beta": VAE_BETA
}

with open(output_path("project_config.json"), "w", encoding="utf-8") as f:
    json.dump(config, f, indent=4)

with open(output_path("project_config.yaml"), "w", encoding="utf-8") as f:
    yaml.safe_dump(config, f, sort_keys=False)

print("Configuration files saved.")


# =====================================================================
# 3. ROBUST MNIST LOADER
# =====================================================================

MNIST_PATH = output_path("mnist.npz")


def validate_mnist(filepath):
    try:
        with np.load(filepath, allow_pickle=True) as data:
            required = {"x_train", "y_train", "x_test", "y_test"}

            if not required.issubset(set(data.files)):
                return False

            if data["x_train"].shape != (60000, 28, 28):
                return False

            if data["x_test"].shape != (10000, 28, 28):
                return False

        return True

    except Exception:
        return False


def load_npz_mnist(filepath):
    with np.load(filepath, allow_pickle=True) as data:
        return (
            data["x_train"],
            data["y_train"],
            data["x_test"],
            data["y_test"]
        )


def load_mnist_robust():

    # ---------------------------------------------------------------
    # Method 1: project folder
    # ---------------------------------------------------------------

    if os.path.exists(MNIST_PATH) and validate_mnist(MNIST_PATH):
        print("\nUsing local MNIST:")
        print(MNIST_PATH)
        return load_npz_mnist(MNIST_PATH)

    # ---------------------------------------------------------------
    # Method 2: Keras cache
    # ---------------------------------------------------------------

    keras_cache = os.path.join(
        os.path.expanduser("~"),
        ".keras",
        "datasets",
        "mnist.npz"
    )

    if os.path.exists(keras_cache) and validate_mnist(keras_cache):
        print("\nUsing MNIST from Keras cache.")

        shutil.copy2(
            keras_cache,
            MNIST_PATH
        )

        return load_npz_mnist(MNIST_PATH)

    # ---------------------------------------------------------------
    # Method 3: normal Keras downloader
    # ---------------------------------------------------------------

    try:
        print("\nTrying Keras MNIST download...")

        (x_train, y_train), (x_test, y_test) = (
            keras.datasets.mnist.load_data()
        )

        np.savez_compressed(
            MNIST_PATH,
            x_train=x_train,
            y_train=y_train,
            x_test=x_test,
            y_test=y_test
        )

        return x_train, y_train, x_test, y_test

    except Exception as e:
        print("Keras download failed:")
        print(e)

    # ---------------------------------------------------------------
    # Method 4: requests
    # ---------------------------------------------------------------

    urls = [
        "https://storage.googleapis.com/tensorflow/tf-keras-datasets/mnist.npz",
        "https://s3.amazonaws.com/img-datasets/mnist.npz"
    ]

    try:
        import requests

        for url in urls:
            try:
                print("\nTrying direct download:")
                print(url)

                response = requests.get(
                    url,
                    headers={"User-Agent": "Mozilla/5.0"},
                    timeout=120
                )

                response.raise_for_status()

                with open(MNIST_PATH, "wb") as f:
                    f.write(response.content)

                if validate_mnist(MNIST_PATH):
                    print("MNIST downloaded successfully.")
                    return load_npz_mnist(MNIST_PATH)

            except Exception as e:
                print("Download failed:")
                print(e)

    except ImportError:
        print("requests is not installed.")

    # ---------------------------------------------------------------
    # Method 5: urllib
    # ---------------------------------------------------------------

    for url in urls:
        try:
            print("\nTrying urllib:")
            print(url)

            request = urllib.request.Request(
                url,
                headers={"User-Agent": "Mozilla/5.0"}
            )

            context = ssl.create_default_context()

            with urllib.request.urlopen(
                request,
                timeout=120,
                context=context
            ) as response:
                with open(MNIST_PATH, "wb") as f:
                    shutil.copyfileobj(response, f)

            if validate_mnist(MNIST_PATH):
                print("MNIST downloaded successfully.")
                return load_npz_mnist(MNIST_PATH)

        except Exception as e:
            print("urllib failed:")
            print(e)

    raise RuntimeError(
        "\nMNIST could not be downloaded.\n"
        "Manually download mnist.npz and place it here:\n"
        + MNIST_PATH
    )


# =====================================================================
# 4. LOAD DATA
# =====================================================================

print("\nLoading MNIST...")

x_train, y_train, x_test, y_test = load_mnist_robust()

print("Training images:", x_train.shape)
print("Training labels:", y_train.shape)
print("Testing images:", x_test.shape)
print("Testing labels:", y_test.shape)


# =====================================================================
# 5. NORMALIZATION
# =====================================================================

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

x_train_flat = x_train.reshape(-1, 784)
x_test_flat = x_test.reshape(-1, 784)

x_train_conv = np.expand_dims(x_train, axis=-1)
x_test_conv = np.expand_dims(x_test, axis=-1)

print("\nFlat shape:", x_train_flat.shape)
print("CNN shape:", x_train_conv.shape)


# =====================================================================
# 6. MNIST SAMPLE VISUALIZATION
# =====================================================================

plt.figure(figsize=(12, 5))

for i in range(20):
    plt.subplot(2, 10, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.title(str(y_train[i]))
    plt.axis("off")

plt.suptitle("MNIST Sample Images")
plt.tight_layout()

plt.savefig(
    output_path("01_mnist_samples.png"),
    dpi=120,
    bbox_inches="tight"
)

plt.show()
plt.close()


# =====================================================================
# 7. NOISE FUNCTIONS
# =====================================================================

def add_gaussian_noise(images, noise_factor=0.5):

    noise = np.random.normal(
        0.0,
        1.0,
        images.shape
    ).astype("float32")

    noisy = images + noise_factor * noise

    return np.clip(
        noisy,
        0.0,
        1.0
    ).astype("float32")


def add_salt_pepper_noise(images, amount=0.1):

    noisy = images.copy()

    random_values = np.random.random(
        images.shape
    )

    noisy[random_values < amount / 2] = 0.0

    noisy[
        random_values > (1.0 - amount / 2)
    ] = 1.0

    return noisy.astype("float32")


# =====================================================================
# 8. CREATE NOISY DATA
# =====================================================================

print("\nCreating noisy data...")

x_train_gaussian = add_gaussian_noise(
    x_train_conv,
    0.5
)

x_test_gaussian = add_gaussian_noise(
    x_test_conv,
    0.5
)

x_train_saltpepper = add_salt_pepper_noise(
    x_train_conv,
    0.1
)

x_test_saltpepper = add_salt_pepper_noise(
    x_test_conv,
    0.1
)


# =====================================================================
# 9. ORIGINAL VS NOISY
# =====================================================================

plt.figure(figsize=(15, 4))

for i in range(10):

    plt.subplot(2, 10, i + 1)
    plt.imshow(x_test[i], cmap="gray")
    plt.axis("off")

    if i == 0:
        plt.ylabel("Original")

    plt.subplot(2, 10, i + 11)
    plt.imshow(
        x_test_gaussian[i].squeeze(),
        cmap="gray"
    )
    plt.axis("off")

    if i == 0:
        plt.ylabel("Noisy")

plt.suptitle("Original vs Gaussian Noisy Images")
plt.tight_layout()

plt.savefig(
    output_path("02_original_vs_noisy.png"),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 10. CALLBACKS
# =====================================================================

def get_callbacks():

    return [
        keras.callbacks.EarlyStopping(
            monitor="val_loss",
            patience=5,
            restore_best_weights=True,
            verbose=1
        ),

        keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss",
            factor=0.5,
            patience=3,
            min_lr=1e-6,
            verbose=1
        )
    ]


# =====================================================================
# 11. VANILLA AUTOENCODER
# =====================================================================

print("\n" + "=" * 70)
print("VANILLA AUTOENCODER")
print("=" * 70)

vanilla_input = keras.Input(
    shape=(784,),
    name="vanilla_input"
)

x = layers.Dense(
    512,
    activation="relu"
)(vanilla_input)

x = layers.Dense(
    256,
    activation="relu"
)(x)

x = layers.Dense(
    128,
    activation="relu"
)(x)

vanilla_latent = layers.Dense(
    VANILLA_LATENT_DIM,
    name="vanilla_latent"
)(x)

x = layers.Dense(
    128,
    activation="relu"
)(vanilla_latent)

x = layers.Dense(
    256,
    activation="relu"
)(x)

x = layers.Dense(
    512,
    activation="relu"
)(x)

vanilla_output = layers.Dense(
    784,
    activation="sigmoid",
    name="vanilla_output"
)(x)

vanilla_autoencoder = keras.Model(
    vanilla_input,
    vanilla_output,
    name="vanilla_autoencoder"
)

vanilla_encoder = keras.Model(
    vanilla_input,
    vanilla_latent,
    name="vanilla_encoder"
)

vanilla_autoencoder.compile(
    optimizer=keras.optimizers.Adam(
        learning_rate=LEARNING_RATE
    ),
    loss="mse"
)

vanilla_autoencoder.summary()


# =====================================================================
# 12. TRAIN VANILLA
# =====================================================================

print("\nTraining Vanilla Autoencoder...")

start_time = time.time()

vanilla_history = vanilla_autoencoder.fit(
    x_train_flat,
    x_train_flat,
    epochs=MAX_EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=VALIDATION_SPLIT,
    shuffle=True,
    callbacks=get_callbacks(),
    verbose=1
)

vanilla_time = time.time() - start_time

print(
    f"\nVanilla training time: "
    f"{vanilla_time / 60:.2f} minutes"
)


# =====================================================================
# 13. SAVE VANILLA
# =====================================================================

vanilla_autoencoder.save_weights(
    output_path(
        "vanilla_autoencoder.weights.h5"
    )
)

vanilla_encoder.save_weights(
    output_path(
        "vanilla_encoder.weights.h5"
    )
)

with open(
    output_path("vanilla_history.pkl"),
    "wb"
) as f:
    pickle.dump(
        vanilla_history.history,
        f
    )


# =====================================================================
# 14. VANILLA LOSS GRAPH
# =====================================================================

plt.figure(figsize=(8, 5))

plt.plot(
    vanilla_history.history["loss"],
    label="Training Loss"
)

plt.plot(
    vanilla_history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("MSE")
plt.title("Vanilla Autoencoder Training Loss")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    output_path("03_vanilla_training_loss.png"),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 15. VANILLA PREDICTIONS
# =====================================================================

vanilla_predictions = vanilla_autoencoder.predict(
    x_test_flat,
    batch_size=512,
    verbose=0
)

vanilla_mse = float(
    np.mean(
        np.square(
            x_test_flat - vanilla_predictions
        )
    )
)

print(
    f"\nVanilla Test MSE: "
    f"{vanilla_mse:.6f}"
)


# =====================================================================
# 16. VANILLA RECONSTRUCTION GRID
# =====================================================================

plt.figure(figsize=(15, 15))

for digit in range(10):

    indices = np.where(
        y_test == digit
    )[0][:5]

    for j in range(5):

        idx = indices[j]

        plt.subplot(
            10,
            10,
            digit * 10 + j * 2 + 1
        )

        plt.imshow(
            x_test[idx],
            cmap="gray"
        )

        plt.axis("off")

        plt.subplot(
            10,
            10,
            digit * 10 + j * 2 + 2
        )

        plt.imshow(
            vanilla_predictions[idx].reshape(
                28,
                28
            ),
            cmap="gray"
        )

        plt.axis("off")

plt.suptitle(
    "Vanilla Autoencoder: Original vs Reconstruction"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "04_vanilla_reconstruction_grid.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 17. VANILLA ERROR PER DIGIT
# =====================================================================

class_errors = []

for digit in range(10):

    mask = y_test == digit

    error = np.mean(
        np.square(
            x_test_flat[mask]
            -
            vanilla_predictions[mask]
        )
    )

    class_errors.append(
        float(error)
    )


error_df = pd.DataFrame({
    "Digit": range(10),
    "MSE": class_errors
})

error_df.to_csv(
    output_path(
        "vanilla_reconstruction_error_by_digit.csv"
    ),
    index=False
)


plt.figure(figsize=(8, 5))

plt.bar(
    error_df["Digit"],
    error_df["MSE"]
)

plt.xlabel("Digit")
plt.ylabel("MSE")
plt.title(
    "Vanilla Reconstruction Error per Digit"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "05_vanilla_error_by_digit.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 18. ERROR HEATMAP
# =====================================================================

heatmap_data = np.array(
    class_errors
).reshape(
    2,
    5
)

plt.figure(figsize=(8, 4))

img = plt.imshow(
    heatmap_data,
    aspect="auto"
)

plt.colorbar(
    img,
    label="MSE"
)

for row in range(2):
    for col in range(5):

        digit = row * 5 + col

        plt.text(
            col,
            row,
            f"Digit {digit}\n"
            f"{heatmap_data[row, col]:.4f}",
            ha="center",
            va="center"
        )

plt.xticks([])
plt.yticks([])
plt.title("Reconstruction Error Heatmap")
plt.tight_layout()

plt.savefig(
    output_path(
        "06_reconstruction_error_heatmap.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 19. VANILLA LATENT SPACE
# =====================================================================

vanilla_latent_vectors = vanilla_encoder.predict(
    x_test_flat,
    batch_size=512,
    verbose=0
)

sample_count = min(
    LATENT_VIS_SAMPLES,
    len(x_test)
)

sample_indices = np.random.choice(
    len(x_test),
    sample_count,
    replace=False
)

latent_sample = vanilla_latent_vectors[
    sample_indices
]

label_sample = y_test[
    sample_indices
]


# =====================================================================
# 20. PCA
# =====================================================================

vanilla_pca = PCA(
    n_components=2
).fit_transform(
    latent_sample
)

plt.figure(figsize=(9, 7))

scatter = plt.scatter(
    vanilla_pca[:, 0],
    vanilla_pca[:, 1],
    c=label_sample,
    s=8,
    alpha=0.7
)

plt.colorbar(
    scatter,
    ticks=range(10)
)

plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.title(
    "Vanilla Autoencoder Latent Space - PCA"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "07_vanilla_latent_pca.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 21. T-SNE
# =====================================================================

print("\nRunning t-SNE...")

vanilla_tsne = TSNE(
    n_components=2,
    random_state=SEED,
    init="pca",
    learning_rate="auto",
    perplexity=30
).fit_transform(
    latent_sample
)

plt.figure(figsize=(9, 7))

scatter = plt.scatter(
    vanilla_tsne[:, 0],
    vanilla_tsne[:, 1],
    c=label_sample,
    s=8,
    alpha=0.7
)

plt.colorbar(
    scatter,
    ticks=range(10)
)

plt.xlabel("t-SNE Component 1")
plt.ylabel("t-SNE Component 2")

plt.title(
    "Vanilla Autoencoder Latent Space - t-SNE"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "08_vanilla_latent_tsne.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 22. CONVOLUTIONAL AUTOENCODER
# =====================================================================
#
# FIXED SHAPE FLOW:
#
# Encoder:
# 28 -> 14 -> 7 -> 4
#
# Decoder:
# 4 -> 8 -> CROP -> 7 -> 14 -> 28
#
# No output_padding is used.
# =====================================================================

print("\n" + "=" * 70)
print("CONVOLUTIONAL AUTOENCODER")
print("=" * 70)

conv_input = keras.Input(
    shape=(28, 28, 1),
    name="conv_input"
)


# 28 -> 14

c = layers.Conv2D(
    32,
    kernel_size=3,
    strides=2,
    padding="same",
    activation="relu",
    name="encoder_conv1"
)(conv_input)

print(
    "After encoder_conv1:",
    c.shape
)


# 14 -> 7

c = layers.Conv2D(
    64,
    kernel_size=3,
    strides=2,
    padding="same",
    activation="relu",
    name="encoder_conv2"
)(c)

print(
    "After encoder_conv2:",
    c.shape
)


# 7 -> 4

c = layers.Conv2D(
    128,
    kernel_size=3,
    strides=2,
    padding="same",
    activation="relu",
    name="encoder_conv3"
)(c)

print(
    "Encoder convolution output:",
    c.shape
)


c = layers.Flatten(
    name="encoder_flatten"
)(c)

conv_latent = layers.Dense(
    CONV_LATENT_DIM,
    name="conv_latent"
)(c)

print(
    "Latent shape:",
    conv_latent.shape
)


# =====================================================================
# 23. FIXED CONVOLUTIONAL DECODER
# =====================================================================

c = layers.Dense(
    4 * 4 * 128,
    activation="relu",
    name="decoder_dense"
)(conv_latent)

c = layers.Reshape(
    (4, 4, 128),
    name="decoder_reshape"
)(c)

print(
    "Decoder starting shape:",
    c.shape
)


# 4 -> 8

c = layers.Conv2DTranspose(
    64,
    kernel_size=3,
    strides=2,
    padding="same",
    activation="relu",
    name="decoder_deconv1"
)(c)

print(
    "After decoder_deconv1:",
    c.shape
)


# 8 -> 7

c = layers.Cropping2D(
    cropping=(
        (0, 1),
        (0, 1)
    ),
    name="decoder_crop"
)(c)

print(
    "After decoder_crop:",
    c.shape
)


# 7 -> 14

c = layers.Conv2DTranspose(
    32,
    kernel_size=3,
    strides=2,
    padding="same",
    activation="relu",
    name="decoder_deconv2"
)(c)

print(
    "After decoder_deconv2:",
    c.shape
)


# 14 -> 28

conv_output = layers.Conv2DTranspose(
    1,
    kernel_size=3,
    strides=2,
    padding="same",
    activation="sigmoid",
    name="decoder_output"
)(c)

print(
    "Final decoder tensor:",
    conv_output.shape
)


conv_autoencoder = keras.Model(
    conv_input,
    conv_output,
    name="convolutional_autoencoder"
)

conv_encoder = keras.Model(
    conv_input,
    conv_latent,
    name="convolutional_encoder"
)


# =====================================================================
# 24. VERIFY OUTPUT SHAPE
# =====================================================================

print(
    "\nConvolutional output shape:",
    conv_autoencoder.output_shape
)

expected_shape = (
    28,
    28,
    1
)

actual_shape = (
    conv_autoencoder.output_shape[1:]
)

if actual_shape != expected_shape:
    raise ValueError(
        f"Incorrect CNN output shape. "
        f"Got {actual_shape}, "
        f"expected {expected_shape}"
    )

print(
    "CNN decoder verified: 28 x 28 x 1"
)


# =====================================================================
# 25. TRAIN CNN
# =====================================================================

conv_autoencoder.compile(
    optimizer=keras.optimizers.Adam(
        learning_rate=LEARNING_RATE
    ),
    loss="mse"
)

conv_autoencoder.summary()

print(
    "\nTraining Convolutional Autoencoder..."
)

start_time = time.time()

conv_history = conv_autoencoder.fit(
    x_train_conv,
    x_train_conv,
    epochs=MAX_EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=VALIDATION_SPLIT,
    shuffle=True,
    callbacks=get_callbacks(),
    verbose=1
)

conv_time = time.time() - start_time

print(
    f"\nCNN training time: "
    f"{conv_time / 60:.2f} minutes"
)


# =====================================================================
# 26. SAVE CNN
# =====================================================================

conv_autoencoder.save_weights(
    output_path(
        "convolutional_autoencoder.weights.h5"
    )
)

conv_encoder.save_weights(
    output_path(
        "convolutional_encoder.weights.h5"
    )
)

with open(
    output_path("conv_history.pkl"),
    "wb"
) as f:
    pickle.dump(
        conv_history.history,
        f
    )


# =====================================================================
# 27. CNN LOSS
# =====================================================================

plt.figure(figsize=(8, 5))

plt.plot(
    conv_history.history["loss"],
    label="Training Loss"
)

plt.plot(
    conv_history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("MSE")
plt.title(
    "Convolutional Autoencoder Training Loss"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    output_path(
        "09_conv_training_loss.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 28. CNN PREDICTIONS
# =====================================================================

conv_predictions = conv_autoencoder.predict(
    x_test_conv,
    batch_size=512,
    verbose=0
)

conv_mse = float(
    np.mean(
        np.square(
            x_test_conv
            -
            conv_predictions
        )
    )
)

print(
    f"\nConvolutional Test MSE: "
    f"{conv_mse:.6f}"
)


# =====================================================================
# 29. CNN RECONSTRUCTION GRID
# =====================================================================

plt.figure(figsize=(15, 15))

for digit in range(10):

    indices = np.where(
        y_test == digit
    )[0][:5]

    for j in range(5):

        idx = indices[j]

        plt.subplot(
            10,
            10,
            digit * 10 + j * 2 + 1
        )

        plt.imshow(
            x_test[idx],
            cmap="gray"
        )

        plt.axis("off")

        plt.subplot(
            10,
            10,
            digit * 10 + j * 2 + 2
        )

        plt.imshow(
            conv_predictions[idx].squeeze(),
            cmap="gray"
        )

        plt.axis("off")

plt.suptitle(
    "Convolutional Autoencoder Reconstruction"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "10_conv_reconstruction_grid.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 30. CNN FILTERS
# =====================================================================

filters, _ = conv_autoencoder.get_layer(
    "encoder_conv1"
).get_weights()

plt.figure(figsize=(10, 10))

for i in range(
    min(32, filters.shape[-1])
):

    plt.subplot(
        4,
        8,
        i + 1
    )

    plt.imshow(
        filters[:, :, 0, i],
        cmap="gray"
    )

    plt.axis("off")

plt.suptitle(
    "Learned First-Layer CNN Filters"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "11_learned_conv_filters.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 31. CNN LATENT PCA
# =====================================================================

conv_latent_vectors = conv_encoder.predict(
    x_test_conv,
    batch_size=512,
    verbose=0
)

conv_pca = PCA(
    n_components=2
).fit_transform(
    conv_latent_vectors[
        sample_indices
    ]
)

plt.figure(figsize=(9, 7))

scatter = plt.scatter(
    conv_pca[:, 0],
    conv_pca[:, 1],
    c=label_sample,
    s=8,
    alpha=0.7
)

plt.colorbar(
    scatter,
    ticks=range(10)
)

plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")

plt.title(
    "Convolutional Autoencoder Latent Space - PCA"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "12_conv_latent_pca.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 32. CNN FEATURE MAPS
# =====================================================================

feature_model = keras.Model(
    inputs=conv_autoencoder.input,
    outputs=conv_autoencoder.get_layer(
        "encoder_conv1"
    ).output
)

feature_maps = feature_model.predict(
    x_test_conv[:1],
    verbose=0
)

plt.figure(figsize=(10, 10))

for i in range(16):

    plt.subplot(
        4,
        4,
        i + 1
    )

    plt.imshow(
        feature_maps[0, :, :, i],
        cmap="gray"
    )

    plt.axis("off")

plt.suptitle(
    "CNN Encoder Feature Maps"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "13_encoder_feature_maps.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 33. MIXED NOISE TRAINING DATA
# =====================================================================

print(
    "\nPreparing mixed-noise dataset..."
)

split_index = (
    len(x_train_conv) // 2
)

x_train_mixed_noise = np.empty_like(
    x_train_conv
)

x_train_mixed_noise[
    :split_index
] = add_gaussian_noise(
    x_train_conv[:split_index],
    0.5
)

x_train_mixed_noise[
    split_index:
] = add_salt_pepper_noise(
    x_train_conv[split_index:],
    0.1
)

shuffle_indices = np.random.permutation(
    len(x_train_conv)
)

x_train_mixed_noise = (
    x_train_mixed_noise[
        shuffle_indices
    ]
)

x_train_clean_mixed = (
    x_train_conv[
        shuffle_indices
    ]
)


# =====================================================================
# 34. DENOISING AUTOENCODER
# =====================================================================

print("\n" + "=" * 70)
print("DENOISING AUTOENCODER")
print("=" * 70)

denoise_input = keras.Input(
    shape=(28, 28, 1),
    name="denoise_input"
)


# Encoder

d = layers.Conv2D(
    32,
    3,
    strides=2,
    padding="same",
    activation="relu"
)(denoise_input)

d = layers.Conv2D(
    64,
    3,
    strides=2,
    padding="same",
    activation="relu"
)(d)

d = layers.Conv2D(
    128,
    3,
    strides=2,
    padding="same",
    activation="relu"
)(d)

d = layers.Flatten()(d)

d = layers.Dense(
    CONV_LATENT_DIM,
    name="denoise_latent"
)(d)


# Decoder

d = layers.Dense(
    4 * 4 * 128,
    activation="relu"
)(d)

d = layers.Reshape(
    (4, 4, 128)
)(d)


# 4 -> 8

d = layers.Conv2DTranspose(
    64,
    3,
    strides=2,
    padding="same",
    activation="relu"
)(d)


# 8 -> 7

d = layers.Cropping2D(
    cropping=(
        (0, 1),
        (0, 1)
    )
)(d)


# 7 -> 14

d = layers.Conv2DTranspose(
    32,
    3,
    strides=2,
    padding="same",
    activation="relu"
)(d)


# 14 -> 28

denoise_output = layers.Conv2DTranspose(
    1,
    3,
    strides=2,
    padding="same",
    activation="sigmoid"
)(d)

denoising_autoencoder = keras.Model(
    denoise_input,
    denoise_output,
    name="denoising_autoencoder"
)

print(
    "Denoising output shape:",
    denoising_autoencoder.output_shape
)

if denoising_autoencoder.output_shape[1:] != (
    28,
    28,
    1
):
    raise ValueError(
        "Denoising output is not 28x28x1."
    )

denoising_autoencoder.compile(
    optimizer=keras.optimizers.Adam(
        learning_rate=LEARNING_RATE
    ),
    loss="mse"
)

denoising_autoencoder.summary()


# =====================================================================
# 35. TRAIN DENOISING MODEL
# =====================================================================

print(
    "\nTraining Denoising Autoencoder..."
)

start_time = time.time()

denoising_history = denoising_autoencoder.fit(
    x_train_mixed_noise,
    x_train_clean_mixed,
    epochs=MAX_EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=VALIDATION_SPLIT,
    shuffle=True,
    callbacks=get_callbacks(),
    verbose=1
)

denoising_time = (
    time.time() - start_time
)

print(
    f"\nDenoising training time: "
    f"{denoising_time / 60:.2f} minutes"
)


# =====================================================================
# 36. SAVE DENOISING MODEL
# =====================================================================

denoising_autoencoder.save_weights(
    output_path(
        "denoising_autoencoder.weights.h5"
    )
)

with open(
    output_path(
        "denoising_history.pkl"
    ),
    "wb"
) as f:
    pickle.dump(
        denoising_history.history,
        f
    )


# =====================================================================
# 37. DENOISING LOSS
# =====================================================================

plt.figure(figsize=(8, 5))

plt.plot(
    denoising_history.history["loss"],
    label="Training Loss"
)

plt.plot(
    denoising_history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("MSE")

plt.title(
    "Denoising Autoencoder Training Loss"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    output_path(
        "14_denoising_training_loss.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 38. DENOISING PREDICTIONS
# =====================================================================

denoised_predictions = (
    denoising_autoencoder.predict(
        x_test_gaussian,
        batch_size=512,
        verbose=0
    )
)

denoising_mse = float(
    np.mean(
        np.square(
            x_test_conv
            -
            denoised_predictions
        )
    )
)

print(
    f"\nDenoising MSE: "
    f"{denoising_mse:.6f}"
)


# =====================================================================
# 39. DENOISING VISUALIZATION
# =====================================================================

plt.figure(figsize=(15, 6))

for i in range(10):

    plt.subplot(
        3,
        10,
        i + 1
    )

    plt.imshow(
        x_test_gaussian[i].squeeze(),
        cmap="gray"
    )

    plt.axis("off")

    if i == 0:
        plt.ylabel("Noisy")

    plt.subplot(
        3,
        10,
        i + 11
    )

    plt.imshow(
        denoised_predictions[i].squeeze(),
        cmap="gray"
    )

    plt.axis("off")

    if i == 0:
        plt.ylabel("Denoised")

    plt.subplot(
        3,
        10,
        i + 21
    )

    plt.imshow(
        x_test[i],
        cmap="gray"
    )

    plt.axis("off")

    if i == 0:
        plt.ylabel("Original")

plt.suptitle(
    "Noisy -> Denoised -> Original"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "15_denoising_results.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 40. PSNR + SSIM FUNCTION
# =====================================================================

def calculate_metrics(
    original,
    corrupted,
    restored
):

    noisy_psnr = []
    denoised_psnr = []
    noisy_ssim = []
    denoised_ssim = []

    for i in range(len(original)):

        original_image = (
            original[i].squeeze()
        )

        corrupted_image = (
            corrupted[i].squeeze()
        )

        restored_image = (
            restored[i].squeeze()
        )

        noisy_psnr.append(
            peak_signal_noise_ratio(
                original_image,
                corrupted_image,
                data_range=1.0
            )
        )

        denoised_psnr.append(
            peak_signal_noise_ratio(
                original_image,
                restored_image,
                data_range=1.0
            )
        )

        noisy_ssim.append(
            structural_similarity(
                original_image,
                corrupted_image,
                data_range=1.0
            )
        )

        denoised_ssim.append(
            structural_similarity(
                original_image,
                restored_image,
                data_range=1.0
            )
        )

    noisy_psnr_mean = float(
        np.mean(noisy_psnr)
    )

    denoised_psnr_mean = float(
        np.mean(denoised_psnr)
    )

    noisy_ssim_mean = float(
        np.mean(noisy_ssim)
    )

    denoised_ssim_mean = float(
        np.mean(denoised_ssim)
    )

    return {
        "noisy_psnr": noisy_psnr_mean,
        "denoised_psnr": denoised_psnr_mean,
        "psnr_improvement":
            denoised_psnr_mean
            -
            noisy_psnr_mean,
        "noisy_ssim": noisy_ssim_mean,
        "denoised_ssim": denoised_ssim_mean,
        "ssim_improvement":
            denoised_ssim_mean
            -
            noisy_ssim_mean
    }


# =====================================================================
# 41. GAUSSIAN METRICS
# =====================================================================

gaussian_metrics = calculate_metrics(
    x_test_conv,
    x_test_gaussian,
    denoised_predictions
)

print("\nGaussian Noise Metrics:")

for key, value in gaussian_metrics.items():
    print(
        f"{key}: {value:.4f}"
    )


# =====================================================================
# 42. DIFFERENT NOISE LEVELS
# =====================================================================

noise_levels = [
    0.3,
    0.5,
    0.7
]

noise_results = []

for level in noise_levels:

    print(
        f"\nTesting noise level {level}..."
    )

    noisy_test = add_gaussian_noise(
        x_test_conv,
        level
    )

    restored_test = (
        denoising_autoencoder.predict(
            noisy_test,
            batch_size=512,
            verbose=0
        )
    )

    metrics = calculate_metrics(
        x_test_conv,
        noisy_test,
        restored_test
    )

    noise_results.append({
        "noise_level": level,
        **metrics
    })


noise_df = pd.DataFrame(
    noise_results
)

noise_df.to_csv(
    output_path(
        "denoising_metrics.csv"
    ),
    index=False
)

print("\nNoise results:")
print(
    noise_df.to_string(
        index=False
    )
)


# =====================================================================
# 43. PSNR GRAPH
# =====================================================================

plt.figure(figsize=(8, 5))

plt.plot(
    noise_df["noise_level"],
    noise_df["noisy_psnr"],
    marker="o",
    label="Noisy PSNR"
)

plt.plot(
    noise_df["noise_level"],
    noise_df["denoised_psnr"],
    marker="o",
    label="Denoised PSNR"
)

plt.xlabel("Gaussian Noise Factor")
plt.ylabel("PSNR (dB)")
plt.title("PSNR Comparison")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    output_path(
        "16_psnr_comparison.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 44. SSIM GRAPH
# =====================================================================

plt.figure(figsize=(8, 5))

plt.plot(
    noise_df["noise_level"],
    noise_df["noisy_ssim"],
    marker="o",
    label="Noisy SSIM"
)

plt.plot(
    noise_df["noise_level"],
    noise_df["denoised_ssim"],
    marker="o",
    label="Denoised SSIM"
)

plt.xlabel("Gaussian Noise Factor")
plt.ylabel("SSIM")
plt.title("SSIM Comparison")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    output_path(
        "17_ssim_comparison.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 45. NOISE LEVEL VISUALIZATION
# =====================================================================

plt.figure(figsize=(15, 9))

plot_index = 1

for level in noise_levels:

    noisy_test = add_gaussian_noise(
        x_test_conv[:5],
        level
    )

    restored_test = (
        denoising_autoencoder.predict(
            noisy_test,
            verbose=0
        )
    )

    for i in range(5):

        plt.subplot(
            6,
            5,
            plot_index
        )

        plt.imshow(
            noisy_test[i].squeeze(),
            cmap="gray"
        )

        plt.axis("off")

        if i == 0:
            plt.ylabel(
                f"Noise {level}"
            )

        plot_index += 1

    for i in range(5):

        plt.subplot(
            6,
            5,
            plot_index
        )

        plt.imshow(
            restored_test[i].squeeze(),
            cmap="gray"
        )

        plt.axis("off")

        if i == 0:
            plt.ylabel("Denoised")

        plot_index += 1


plt.suptitle(
    "Denoising at Noise Levels 0.3, 0.5 and 0.7"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "18_noise_level_denoising.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 46. SALT AND PEPPER
# =====================================================================

saltpepper_predictions = (
    denoising_autoencoder.predict(
        x_test_saltpepper,
        batch_size=512,
        verbose=0
    )
)

saltpepper_metrics = calculate_metrics(
    x_test_conv,
    x_test_saltpepper,
    saltpepper_predictions
)

print(
    "\nSalt-and-Pepper Metrics:"
)

for key, value in saltpepper_metrics.items():
    print(
        f"{key}: {value:.4f}"
    )


# =====================================================================
# 47. SALT-PEPPER VISUALIZATION
# =====================================================================

plt.figure(figsize=(15, 6))

for i in range(10):

    plt.subplot(
        3,
        10,
        i + 1
    )

    plt.imshow(
        x_test_saltpepper[i].squeeze(),
        cmap="gray"
    )

    plt.axis("off")

    plt.subplot(
        3,
        10,
        i + 11
    )

    plt.imshow(
        saltpepper_predictions[i].squeeze(),
        cmap="gray"
    )

    plt.axis("off")

    plt.subplot(
        3,
        10,
        i + 21
    )

    plt.imshow(
        x_test[i],
        cmap="gray"
    )

    plt.axis("off")

plt.suptitle(
    "Salt-and-Pepper: Noisy -> Denoised -> Original"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "19_salt_pepper_denoising.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 48. VANILLA VS CNN
# =====================================================================

plt.figure(figsize=(15, 6))

for i in range(10):

    plt.subplot(
        3,
        10,
        i + 1
    )

    plt.imshow(
        x_test[i],
        cmap="gray"
    )

    plt.axis("off")

    plt.subplot(
        3,
        10,
        i + 11
    )

    plt.imshow(
        vanilla_predictions[i].reshape(
            28,
            28
        ),
        cmap="gray"
    )

    plt.axis("off")

    plt.subplot(
        3,
        10,
        i + 21
    )

    plt.imshow(
        conv_predictions[i].squeeze(),
        cmap="gray"
    )

    plt.axis("off")

plt.suptitle(
    "Original vs Vanilla vs Convolutional"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "20_vanilla_vs_conv_reconstruction.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 49. VAE SAMPLING LAYER
# =====================================================================

class Sampling(layers.Layer):

    def call(self, inputs):

        z_mean, z_log_var = inputs

        epsilon = tf.random.normal(
            shape=tf.shape(z_mean)
        )

        return (
            z_mean
            +
            tf.exp(
                0.5 * z_log_var
            )
            *
            epsilon
        )


# =====================================================================
# 50. VAE ENCODER
# =====================================================================

print("\n" + "=" * 70)
print("VARIATIONAL AUTOENCODER")
print("=" * 70)

vae_input = keras.Input(
    shape=(28, 28, 1),
    name="vae_input"
)

v = layers.Conv2D(
    32,
    3,
    strides=2,
    padding="same",
    activation="relu"
)(vae_input)

v = layers.Conv2D(
    64,
    3,
    strides=2,
    padding="same",
    activation="relu"
)(v)

v = layers.Flatten()(v)

v = layers.Dense(
    128,
    activation="relu"
)(v)

z_mean = layers.Dense(
    VAE_LATENT_DIM,
    name="z_mean"
)(v)

z_log_var = layers.Dense(
    VAE_LATENT_DIM,
    name="z_log_var"
)(v)

z_sample = Sampling()(
    [
        z_mean,
        z_log_var
    ]
)

vae_encoder = keras.Model(
    vae_input,
    [
        z_mean,
        z_log_var,
        z_sample
    ],
    name="vae_encoder"
)


# =====================================================================
# 51. VAE DECODER
# =====================================================================

vae_latent_input = keras.Input(
    shape=(VAE_LATENT_DIM,),
    name="vae_latent_input"
)

v = layers.Dense(
    7 * 7 * 64,
    activation="relu"
)(vae_latent_input)

v = layers.Reshape(
    (7, 7, 64)
)(v)

v = layers.Conv2DTranspose(
    64,
    3,
    strides=2,
    padding="same",
    activation="relu"
)(v)

v = layers.Conv2DTranspose(
    32,
    3,
    strides=2,
    padding="same",
    activation="relu"
)(v)

vae_output = layers.Conv2D(
    1,
    3,
    padding="same",
    activation="sigmoid"
)(v)

vae_decoder = keras.Model(
    vae_latent_input,
    vae_output,
    name="vae_decoder"
)

print(
    "VAE decoder output:",
    vae_decoder.output_shape
)


# =====================================================================
# 52. CUSTOM VAE
# =====================================================================

class VAE(keras.Model):

    def __init__(
        self,
        encoder,
        decoder,
        beta=1.0,
        **kwargs
    ):

        super().__init__(**kwargs)

        self.encoder = encoder
        self.decoder = decoder
        self.beta = beta

        self.total_loss_tracker = (
            keras.metrics.Mean(
                name="loss"
            )
        )

        self.reconstruction_loss_tracker = (
            keras.metrics.Mean(
                name="reconstruction_loss"
            )
        )

        self.kl_loss_tracker = (
            keras.metrics.Mean(
                name="kl_loss"
            )
        )

    @property
    def metrics(self):

        return [
            self.total_loss_tracker,
            self.reconstruction_loss_tracker,
            self.kl_loss_tracker
        ]

    def train_step(self, data):

        if isinstance(data, tuple):
            data = data[0]

        with tf.GradientTape() as tape:

            current_mean, current_log_var, current_z = (
                self.encoder(
                    data,
                    training=True
                )
            )

            reconstruction = self.decoder(
                current_z,
                training=True
            )

            reconstruction_loss = tf.reduce_mean(
                tf.reduce_sum(
                    tf.square(
                        data - reconstruction
                    ),
                    axis=(1, 2, 3)
                )
            )

            kl_loss = -0.5 * tf.reduce_mean(
                tf.reduce_sum(
                    1
                    +
                    current_log_var
                    -
                    tf.square(
                        current_mean
                    )
                    -
                    tf.exp(
                        current_log_var
                    ),
                    axis=1
                )
            )

            total_loss = (
                reconstruction_loss
                +
                self.beta * kl_loss
            )

        gradients = tape.gradient(
            total_loss,
            self.trainable_weights
        )

        self.optimizer.apply_gradients(
            zip(
                gradients,
                self.trainable_weights
            )
        )

        self.total_loss_tracker.update_state(
            total_loss
        )

        self.reconstruction_loss_tracker.update_state(
            reconstruction_loss
        )

        self.kl_loss_tracker.update_state(
            kl_loss
        )

        return {
            "loss":
                self.total_loss_tracker.result(),

            "reconstruction_loss":
                self.reconstruction_loss_tracker.result(),

            "kl_loss":
                self.kl_loss_tracker.result()
        }


# =====================================================================
# 53. TRAIN VAE
# =====================================================================

vae = VAE(
    vae_encoder,
    vae_decoder,
    beta=VAE_BETA,
    name="vae"
)

vae.compile(
    optimizer=keras.optimizers.Adam(
        learning_rate=LEARNING_RATE
    )
)

print("\nTraining VAE...")

start_time = time.time()

vae_history = vae.fit(
    x_train_conv,
    epochs=MAX_EPOCHS,
    batch_size=BATCH_SIZE,
    shuffle=True,
    verbose=1
)

vae_time = time.time() - start_time

print(
    f"\nVAE training time: "
    f"{vae_time / 60:.2f} minutes"
)


# =====================================================================
# 54. SAVE VAE
# =====================================================================

vae_encoder.save_weights(
    output_path(
        "vae_encoder.weights.h5"
    )
)

vae_decoder.save_weights(
    output_path(
        "vae_decoder.weights.h5"
    )
)

with open(
    output_path(
        "vae_history.pkl"
    ),
    "wb"
) as f:
    pickle.dump(
        vae_history.history,
        f
    )


# =====================================================================
# 55. VAE LOSS GRAPH
# =====================================================================

plt.figure(figsize=(8, 5))

plt.plot(
    vae_history.history["loss"],
    label="Total Loss"
)

if "reconstruction_loss" in vae_history.history:

    plt.plot(
        vae_history.history[
            "reconstruction_loss"
        ],
        label="Reconstruction Loss"
    )

if "kl_loss" in vae_history.history:

    plt.plot(
        vae_history.history["kl_loss"],
        label="KL Loss"
    )

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("VAE Training Loss")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    output_path(
        "21_vae_training_loss.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 56. VAE LATENT SPACE
# =====================================================================

vae_mean, vae_log_var, vae_z = (
    vae_encoder.predict(
        x_test_conv,
        batch_size=512,
        verbose=0
    )
)

vae_pca = PCA(
    n_components=2
).fit_transform(
    vae_mean[
        sample_indices
    ]
)

plt.figure(figsize=(9, 7))

scatter = plt.scatter(
    vae_pca[:, 0],
    vae_pca[:, 1],
    c=label_sample,
    s=8,
    alpha=0.7
)

plt.colorbar(
    scatter,
    ticks=range(10)
)

plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.title("VAE Latent Space - PCA")

plt.tight_layout()

plt.savefig(
    output_path(
        "22_vae_latent_space.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 57. GENERATED VAE DIGITS
# =====================================================================

random_z = np.random.normal(
    size=(
        25,
        VAE_LATENT_DIM
    )
).astype("float32")

generated_digits = vae_decoder.predict(
    random_z,
    verbose=0
)

plt.figure(figsize=(8, 8))

for i in range(25):

    plt.subplot(
        5,
        5,
        i + 1
    )

    plt.imshow(
        generated_digits[i].squeeze(),
        cmap="gray"
    )

    plt.axis("off")

plt.suptitle(
    "VAE Generated Samples"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "23_vae_generated_samples.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 58. VAE LATENT INTERPOLATION
# =====================================================================

digit_a = 1
digit_b = 8

index_a = np.where(
    y_test == digit_a
)[0][0]

index_b = np.where(
    y_test == digit_b
)[0][0]

vector_a = vae_mean[
    index_a
]

vector_b = vae_mean[
    index_b
]

interpolation_vectors = []

for alpha in np.linspace(
    0,
    1,
    12
):

    vector = (
        (1 - alpha) * vector_a
        +
        alpha * vector_b
    )

    interpolation_vectors.append(
        vector
    )

interpolation_vectors = np.array(
    interpolation_vectors,
    dtype="float32"
)

interpolated_images = (
    vae_decoder.predict(
        interpolation_vectors,
        verbose=0
    )
)

plt.figure(figsize=(15, 2))

for i in range(12):

    plt.subplot(
        1,
        12,
        i + 1
    )

    plt.imshow(
        interpolated_images[i].squeeze(),
        cmap="gray"
    )

    plt.axis("off")

plt.suptitle(
    f"VAE Latent Interpolation: "
    f"{digit_a} -> {digit_b}"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "24_vae_latent_interpolation.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 59. VAE MANIFOLD
# =====================================================================

grid_size = 10

grid_values = np.linspace(
    -3,
    3,
    grid_size
)

canvas = np.zeros(
    (
        28 * grid_size,
        28 * grid_size
    )
)

for row, value_y in enumerate(
    grid_values
):

    for col, value_x in enumerate(
        grid_values
    ):

        z_point = np.zeros(
            (
                1,
                VAE_LATENT_DIM
            ),
            dtype="float32"
        )

        z_point[0, 0] = value_x
        z_point[0, 1] = value_y

        generated = vae_decoder.predict(
            z_point,
            verbose=0
        )[0].squeeze()

        canvas[
            row * 28:
            (row + 1) * 28,

            col * 28:
            (col + 1) * 28
        ] = generated


plt.figure(figsize=(10, 10))

plt.imshow(
    canvas,
    cmap="gray"
)

plt.axis("off")
plt.title("VAE Digit Manifold")
plt.tight_layout()

plt.savefig(
    output_path(
        "25_vae_digit_manifold.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 60. VAE RECONSTRUCTION MSE
# =====================================================================

vae_reconstructed = vae_decoder.predict(
    vae_mean,
    batch_size=512,
    verbose=0
)

vae_mse = float(
    np.mean(
        np.square(
            x_test_conv
            -
            vae_reconstructed
        )
    )
)

print(
    f"\nVAE Reconstruction MSE: "
    f"{vae_mse:.6f}"
)


# =====================================================================
# 61. COMPRESSION RATIOS
# =====================================================================

ORIGINAL_DIMENSION = 784

vanilla_ratio = (
    ORIGINAL_DIMENSION
    /
    VANILLA_LATENT_DIM
)

conv_ratio = (
    ORIGINAL_DIMENSION
    /
    CONV_LATENT_DIM
)

vae_ratio = (
    ORIGINAL_DIMENSION
    /
    VAE_LATENT_DIM
)

print("\nCompression Ratios:")

print(
    f"Vanilla: "
    f"{vanilla_ratio:.2f}:1"
)

print(
    f"Convolutional: "
    f"{conv_ratio:.2f}:1"
)

print(
    f"VAE: "
    f"{vae_ratio:.2f}:1"
)


# =====================================================================
# 62. COMPRESSION QUALITY TABLE
# =====================================================================

compression_df = pd.DataFrame({
    "Model": [
        "Vanilla",
        "Convolutional",
        "VAE"
    ],

    "Latent_Dimension": [
        VANILLA_LATENT_DIM,
        CONV_LATENT_DIM,
        VAE_LATENT_DIM
    ],

    "Compression_Ratio": [
        vanilla_ratio,
        conv_ratio,
        vae_ratio
    ],

    "Reconstruction_MSE": [
        vanilla_mse,
        conv_mse,
        vae_mse
    ]
})

compression_df.to_csv(
    output_path(
        "compression_quality_comparison.csv"
    ),
    index=False
)


# =====================================================================
# 63. COMPRESSION VS QUALITY
# =====================================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    compression_df[
        "Compression_Ratio"
    ],
    compression_df[
        "Reconstruction_MSE"
    ],
    s=100
)

for _, row in compression_df.iterrows():

    plt.annotate(
        row["Model"],
        (
            row["Compression_Ratio"],
            row["Reconstruction_MSE"]
        )
    )

plt.xlabel("Compression Ratio")
plt.ylabel("Reconstruction MSE")

plt.title(
    "Compression Ratio vs Reconstruction Quality"
)

plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    output_path(
        "26_compression_vs_quality.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 64. MODEL COMPARISON
# =====================================================================

comparison_df = pd.DataFrame({
    "Model": [
        "Vanilla",
        "Convolutional",
        "Denoising",
        "VAE"
    ],

    "Reconstruction_MSE": [
        vanilla_mse,
        conv_mse,
        denoising_mse,
        vae_mse
    ]
})

comparison_df.to_csv(
    output_path(
        "model_reconstruction_comparison.csv"
    ),
    index=False
)

plt.figure(figsize=(8, 5))

plt.bar(
    comparison_df["Model"],
    comparison_df["Reconstruction_MSE"]
)

plt.ylabel("Reconstruction MSE")

plt.title(
    "Autoencoder Reconstruction Comparison"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "27_model_loss_comparison.png"
    ),
    dpi=120
)

plt.show()
plt.close()


# =====================================================================
# 65. FINAL COMPARATIVE ANALYSIS
# =====================================================================

performance_df = pd.DataFrame({
    "Model": [
        "Vanilla",
        "Convolutional",
        "Denoising",
        "VAE"
    ],

    "Latent_Dim": [
        VANILLA_LATENT_DIM,
        CONV_LATENT_DIM,
        CONV_LATENT_DIM,
        VAE_LATENT_DIM
    ],

    "Reconstruction_Loss": [
        vanilla_mse,
        conv_mse,
        denoising_mse,
        vae_mse
    ],

    "Training_Time_Min": [
        vanilla_time / 60,
        conv_time / 60,
        denoising_time / 60,
        vae_time / 60
    ],

    "Best_Use_Case": [
        "Basic compression",
        "Image compression",
        "Image denoising",
        "Generative modeling"
    ]
})

performance_df.to_csv(
    output_path(
        "comparative_analysis_table.csv"
    ),
    index=False
)

print("\nComparative Analysis:")
print(
    performance_df.to_string(
        index=False
    )
)


# =====================================================================
# 66. SAVE FINAL METRICS
# =====================================================================

final_metrics = {
    "Vanilla": {
        "latent_dimension":
            VANILLA_LATENT_DIM,

        "MSE":
            vanilla_mse,

        "training_minutes":
            vanilla_time / 60,

        "compression_ratio":
            vanilla_ratio
    },

    "Convolutional": {
        "latent_dimension":
            CONV_LATENT_DIM,

        "MSE":
            conv_mse,

        "training_minutes":
            conv_time / 60,

        "compression_ratio":
            conv_ratio
    },

    "Denoising": {
        "latent_dimension":
            CONV_LATENT_DIM,

        "MSE":
            denoising_mse,

        "training_minutes":
            denoising_time / 60,

        "gaussian_metrics":
            gaussian_metrics,

        "salt_pepper_metrics":
            saltpepper_metrics
    },

    "VAE": {
        "latent_dimension":
            VAE_LATENT_DIM,

        "MSE":
            vae_mse,

        "training_minutes":
            vae_time / 60,

        "compression_ratio":
            vae_ratio,

        "beta":
            VAE_BETA
    }
}

with open(
    output_path(
        "final_metrics.json"
    ),
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        final_metrics,
        f,
        indent=4
    )


with open(
    output_path(
        "final_metrics.yaml"
    ),
    "w",
    encoding="utf-8"
) as f:

    yaml.safe_dump(
        final_metrics,
        f,
        sort_keys=False
    )


# =====================================================================
# 67. PROJECT SUMMARY
# =====================================================================

summary_text = f"""
AUTOENCODERS FOR DIMENSIONALITY REDUCTION
=========================================

Dataset:
MNIST

Training Samples:
{len(x_train)}

Testing Samples:
{len(x_test)}

Image Size:
28 x 28

Original Dimension:
784


VANILLA AUTOENCODER
-------------------
Latent Dimension: {VANILLA_LATENT_DIM}
MSE: {vanilla_mse:.6f}
Compression Ratio: {vanilla_ratio:.2f}:1
Training Time: {vanilla_time / 60:.2f} minutes


CONVOLUTIONAL AUTOENCODER
-------------------------
Latent Dimension: {CONV_LATENT_DIM}
MSE: {conv_mse:.6f}
Compression Ratio: {conv_ratio:.2f}:1
Training Time: {conv_time / 60:.2f} minutes


DENOISING AUTOENCODER
---------------------
Latent Dimension: {CONV_LATENT_DIM}
MSE: {denoising_mse:.6f}
Training Time: {denoising_time / 60:.2f} minutes

Gaussian Noisy PSNR:
{gaussian_metrics["noisy_psnr"]:.4f} dB

Gaussian Denoised PSNR:
{gaussian_metrics["denoised_psnr"]:.4f} dB

PSNR Improvement:
{gaussian_metrics["psnr_improvement"]:.4f} dB

Gaussian Noisy SSIM:
{gaussian_metrics["noisy_ssim"]:.4f}

Gaussian Denoised SSIM:
{gaussian_metrics["denoised_ssim"]:.4f}

SSIM Improvement:
{gaussian_metrics["ssim_improvement"]:.4f}


VARIATIONAL AUTOENCODER
-----------------------
Latent Dimension: {VAE_LATENT_DIM}
Beta: {VAE_BETA}
MSE: {vae_mse:.6f}
Compression Ratio: {vae_ratio:.2f}:1
Training Time: {vae_time / 60:.2f} minutes


KEY FINDINGS
------------

1. Vanilla Autoencoder compresses 784 image features into 32 features.

2. Convolutional Autoencoder compresses the image into a
   16-dimensional latent representation.

3. Convolutional layers preserve spatial image features such as
   strokes, edges and curves.

4. Denoising Autoencoder can reconstruct cleaner images from
   Gaussian and Salt-and-Pepper corrupted images.

5. PSNR and SSIM quantify image restoration performance.

6. Increasing Gaussian noise makes reconstruction more difficult.

7. PCA and t-SNE help visualize the learned latent representations.

8. Smaller bottlenecks provide stronger compression but may reduce
   reconstruction quality.

9. The VAE learns a probabilistic latent representation.

10. VAE latent vectors can be sampled to generate new digit images.

11. Latent interpolation demonstrates smooth transitions between
    learned digit representations.
"""

with open(
    output_path(
        "project_summary.txt"
    ),
    "w",
    encoding="utf-8"
) as f:

    f.write(summary_text)


# =====================================================================
# 68. FILE INVENTORY
# =====================================================================

inventory = []

for filename in os.listdir(
    PROJECT_DIR
):

    filepath = os.path.join(
        PROJECT_DIR,
        filename
    )

    if os.path.isfile(filepath):

        size_mb = (
            os.path.getsize(filepath)
            /
            (1024 ** 2)
        )

        inventory.append({
            "File": filename,
            "Size_MB": round(
                size_mb,
                4
            )
        })


inventory_df = pd.DataFrame(
    inventory
)

inventory_df.to_csv(
    output_path(
        "file_inventory.csv"
    ),
    index=False
)


# =====================================================================
# 69. CREATE SUBMISSION FOLDER
# =====================================================================

SUBMISSION_DIR = os.path.join(
    PROJECT_DIR,
    "submission_files"
)

if os.path.exists(
    SUBMISSION_DIR
):
    shutil.rmtree(
        SUBMISSION_DIR
    )

os.makedirs(
    SUBMISSION_DIR,
    exist_ok=True
)

allowed_extensions = (
    ".png",
    ".csv",
    ".json",
    ".yaml",
    ".yml",
    ".pkl",
    ".txt",
    ".h5"
)

for filename in os.listdir(
    PROJECT_DIR
):

    source = os.path.join(
        PROJECT_DIR,
        filename
    )

    if not os.path.isfile(source):
        continue

    if filename.lower() == "mnist.npz":
        continue

    if filename.lower().endswith(
        allowed_extensions
    ):

        destination = os.path.join(
            SUBMISSION_DIR,
            filename
        )

        shutil.copy2(
            source,
            destination
        )


# =====================================================================
# 70. CREATE ZIP
# =====================================================================

ZIP_BASE = os.path.join(
    os.path.dirname(PROJECT_DIR),
    "Autoencoders_for_Dimensionality_Reduction_Submission"
)

ZIP_FILE = ZIP_BASE + ".zip"

if os.path.exists(ZIP_FILE):
    os.remove(ZIP_FILE)

zip_file = shutil.make_archive(
    ZIP_BASE,
    "zip",
    root_dir=SUBMISSION_DIR
)

zip_size_mb = (
    os.path.getsize(zip_file)
    /
    (1024 ** 2)
)

print(
    f"\nInitial ZIP size: "
    f"{zip_size_mb:.2f} MB"
)


# =====================================================================
# 71. REDUCE ZIP SIZE IF NECESSARY
# =====================================================================

if zip_size_mb > 50:

    print(
        "\nZIP exceeds 50 MB."
    )

    print(
        "Removing optional VAE/encoder weights "
        "from submission ZIP."
    )

    optional_files = [
        "vanilla_encoder.weights.h5",
        "convolutional_encoder.weights.h5",
        "vae_encoder.weights.h5",
        "vae_decoder.weights.h5"
    ]

    for filename in optional_files:

        filepath = os.path.join(
            SUBMISSION_DIR,
            filename
        )

        if os.path.exists(filepath):
            os.remove(filepath)

    if os.path.exists(zip_file):
        os.remove(zip_file)

    zip_file = shutil.make_archive(
        ZIP_BASE,
        "zip",
        root_dir=SUBMISSION_DIR
    )

    zip_size_mb = (
        os.path.getsize(zip_file)
        /
        (1024 ** 2)
    )


# =====================================================================
# 72. FINAL OUTPUT
# =====================================================================

print("\n" + "=" * 70)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print(
    f"\nVanilla MSE: "
    f"{vanilla_mse:.6f}"
)

print(
    f"Convolutional MSE: "
    f"{conv_mse:.6f}"
)

print(
    f"Denoising MSE: "
    f"{denoising_mse:.6f}"
)

print(
    f"VAE MSE: "
    f"{vae_mse:.6f}"
)

print(
    f"\nPSNR Improvement: "
    f"{gaussian_metrics['psnr_improvement']:.4f} dB"
)

print(
    f"SSIM Improvement: "
    f"{gaussian_metrics['ssim_improvement']:.4f}"
)

print("\nProject directory:")
print(PROJECT_DIR)

print("\nSubmission folder:")
print(SUBMISSION_DIR)

print("\nZIP file:")
print(zip_file)

print(
    f"\nZIP size: "
    f"{zip_size_mb:.2f} MB"
)

if zip_size_mb <= 50:
    print(
        "\nZIP is below the 50 MB limit."
    )
else:
    print(
        "\nWARNING: ZIP is still above 50 MB."
    )

print("\nAll processing completed.")
```
